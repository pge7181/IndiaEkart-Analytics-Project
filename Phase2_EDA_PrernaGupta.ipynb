# Monthly Order Volume - a bar or line chart showing number of orders per month for all 24 months. Highlight the diwali months (Oct, Nov)

```python
# Monthly Order Volume - bar chart of number of orders per month (24 months), Diwali months (Oct/Nov) highlighted
import matplotlib.pyplot as plt

# Ensure order_date is datetime (same format used during cleaning)
orders_df['order_date'] = pd.to_datetime(orders_df['order_date'], format='%d-%m-%Y')

# Count orders per month
monthly_orders = orders_df.groupby(orders_df['order_date'].dt.to_period('M')).size().sort_index()
monthly_orders.index = monthly_orders.index.to_timestamp()

# Highlight Diwali months (October & November) in a different color
bar_colors = ['#E67E22' if m.month in (10, 11) else '#5B8DB8' for m in monthly_orders.index]

fig, ax = plt.subplots(figsize=(16, 6))
bars = ax.bar(monthly_orders.index, monthly_orders.values, width=22, color=bar_colors, edgecolor='black', linewidth=0.5)

# Annotate each bar with its order count
for bar, count in zip(bars, monthly_orders.values):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{count}',
            ha='center', va='bottom', fontsize=8)

ax.set_title('Monthly Order Volume (24 months) — Diwali months (Oct, Nov) highlighted in orange', fontsize=14)
ax.set_xlabel('Month')
ax.set_ylabel('Number of Orders')
ax.set_xticks(monthly_orders.index)
ax.set_xticklabels([m.strftime('%b-%y') for m in monthly_orders.index], rotation=45, ha='right')
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print(f"Number of months captured: {monthly_orders.shape[0]}")
print(f"Diwali months (Oct+Nov) avg orders : {monthly_orders[[m.month in (10, 11) for m in monthly_orders.index]].mean():.0f}")
print(f"Non-Diwali months avg orders        : {monthly_orders[[m.month not in (10, 11) for m in monthly_orders.index]].mean():.0f}")
print(f"Peak month : {monthly_orders.idxmax().strftime('%b-%y')} with {monthly_orders.max()} orders")
print(f"Low month  : {monthly_orders.idxmin().strftime('%b-%y')} with {monthly_orders.min()} orders")


```


### ------------------------------ Observations ------------------------------

**1. Seasonality (Diwali spike):** The orange Oct/Nov bars clearly stand out from the rest of the year in both years, confirming a strong Diwali-driven seasonality. The Diwali-month average order volume is substantially higher than the non-Diwali average, and one of the Oct/Nov bars is typically the peak month of the entire 24-month window.

**2. Year-over-year growth/decline:** Comparing the same month across the two years (e.g., Jan-Yr1 vs Jan-Yr2) shows whether the business is growing or shrinking; if the Year-2 bars are consistently taller, order volume is trending upward, and if the Diwali spike in Year 2 is larger than Year 1, the festive effect is strengthening over time.

**3. Trough months:** The quietest months sit well below the annual average, likely in the early part of the calendar year (post-festive slump). These low seasons are candidates for promotions or discount campaigns to smooth demand.


# Monthly Revenue (GMV) - same as above but using final amount. Compare Year-Over-Year.

```python
# Monthly Revenue (GMV) - bar chart of total final_amount per month (24 months), Diwali months (Oct/Nov) highlighted
import matplotlib.pyplot as plt

# Ensure order_date is datetime (same format used during cleaning)
orders_df['order_date'] = pd.to_datetime(orders_df['order_date'], format='%d-%m-%Y')

# Sum final_amount per month (GMV)
monthly_gmv = orders_df.groupby(orders_df['order_date'].dt.to_period('M'))['final_amount'].sum().sort_index()
monthly_gmv.index = monthly_gmv.index.to_timestamp()

# Highlight Diwali months (October & November) in a different color
bar_colors = ['#E67E22' if m.month in (10, 11) else '#5B8DB8' for m in monthly_gmv.index]

fig, ax = plt.subplots(figsize=(16, 6))
bars = ax.bar(monthly_gmv.index, monthly_gmv.values, width=22, color=bar_colors, edgecolor='black', linewidth=0.5)

# Annotate each bar with its GMV (in lakhs)
for bar, gmv in zip(bars, monthly_gmv.values):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{gmv / 100000:.1f}L',
            ha='center', va='bottom', fontsize=8)

ax.set_title('Monthly Revenue (GMV) — Diwali months (Oct, Nov) highlighted in orange', fontsize=14)
ax.set_xlabel('Month')
ax.set_ylabel('GMV (final_amount)')
ax.set_xticks(monthly_gmv.index)
ax.set_xticklabels([m.strftime('%b-%y') for m in monthly_gmv.index], rotation=45, ha='right')
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Year-over-Year comparison
years = sorted(set(monthly_gmv.index.year))
for y in years:
    subset = monthly_gmv[monthly_gmv.index.year == y]
    print(f"{y}: Total GMV = {subset.sum():,.0f} | Monthly avg = {subset.mean():,.0f} | Peak month = {subset.idxmax().strftime('%b-%y')} ({subset.max():,.0f})")

print(f"Number of months captured: {monthly_gmv.shape[0]}")
print(f"Peak GMV month : {monthly_gmv.idxmax().strftime('%b-%y')} with {monthly_gmv.max():,.0f}")
print(f"Lowest GMV month : {monthly_gmv.idxmin().strftime('%b-%y')} with {monthly_gmv.min():,.0f}")

# Year-over-Year growth (compare full-year totals and matched months)
if len(years) >= 2:
    yoy_total_growth = (monthly_gmv[monthly_gmv.index.year == years[-1]].sum() /
                        monthly_gmv[monthly_gmv.index.year == years[0]].sum() - 1) * 100
    print(f"YoY GMV growth ({years[0]} -> {years[-1]}): {yoy_total_growth:+.1f}%")

    # Compare same months across years
    gmv_pivot = monthly_gmv.copy()
    gmv_pivot.index = [m.strftime('%b') for m in gmv_pivot.index]
    common_months = [m for m in gmv_pivot.groupby(level=0).size()[gmv_pivot.groupby(level=0).size() == len(years)].index]
    yr1 = gmv_pivot[gmv_pivot.index.isin(common_months) & (monthly_gmv.index.year == years[0])]
    yr2 = gmv_pivot[gmv_pivot.index.isin(common_months) & (monthly_gmv.index.year == years[-1])]
    matched_growth = (yr2.values / yr1.values - 1) * 100
    print(f"Matched-month YoY growth ({years[0]} -> {years[-1]}, {len(common_months)} months): {matched_growth.mean():+.1f}% avg")
```

### ------------------------------ Observations ------------------------------

**1. Partial-year coverage caveat:** The chart captures 25 monthly bars (not the expected 24), spanning parts of three calendar years — 7 months in 2023, a full 12 in 2024, and 6 in 2025. The -8.4% total YoY figure (2023 → 2025) is therefore not a like-for-like comparison and should not be quoted as a decline; the monthly averages are the fairer metric, and they show a stable-to-rising trend (2023: ~₹118M/mo → 2024: ~₹130M/mo → 2025: ~₹126M/mo).

**2. Matched-month YoY:** Only one month name appears in both endpoint years, so the matched-month growth (+278.3%) is driven by a single pair and is highly sensitive to the low base (June 2023 is the trough at ~₹27M, roughly 5x below the peak of ~₹138M in Jan-25). Comparing consecutive years (2023→2024, 2024→2025) instead would give a more robust month-matched growth figure.

**3. Revenue vs volume divergence:** Monthly GMV is far steadier than order volume — the peak-to-trough GMV spread (~₹27M to ~₹138M) and the near-flat monthly averages suggest high-value orders smooth out revenue across the year. Notably, the GMV peak (Jan-25) and the 2023 peak (Sep-23) do **not** land in the Diwali window, indicating the festive spike is driven by order count rather than order value — customers buy more items per order, not necessarily bigger-ticket items, during Diwali.

# Category-wise revenue share - a pie chart or a horizontal bar chart. Which category earns the most?

```python
# Category-wise revenue share - pie chart + horizontal bar chart of GMV per category
import matplotlib.pyplot as plt

# Ensure order_date is datetime (same format used during cleaning)
orders_df['order_date'] = pd.to_datetime(orders_df['order_date'], format='%d-%m-%Y')

# Join orders with products to get the category for each order
# (orders_df has no product_id — order_items_df is the bridge between orders and products.
#  One order can span multiple categories, so GMV is attributed at the item level.)
orders_cat = (
    orders_df[['order_id']]
    .merge(order_items_df[['order_id', 'product_id', 'total_price']], on='order_id', how='left')
    .merge(products_df[['product_id', 'category']], on='product_id', how='left')
)

# Check for any orders that could not be matched to a category
unmatched = orders_cat['category'].isna().sum()
if unmatched > 0:
    print(f"Warning: {unmatched} order lines have no matching category in products_df")

# Total revenue (GMV) per category
# (final_amount is order-level and spans categories, so item-level total_price is used)
cat_gmv = (
    orders_cat.groupby('category')['total_price']
    .sum()
    .sort_values(ascending=True)  # ascending so largest bar is on top in barh
)
cat_share = (cat_gmv / cat_gmv.sum() * 100).round(1)

# ----- Horizontal bar chart -----
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))

colors = plt.cm.tab20.colors[:len(cat_gmv)]
bars = ax1.barh(cat_gmv.index, cat_gmv.values, color=colors, edgecolor='black', linewidth=0.5)
for bar, gmv, share in zip(bars, cat_gmv.values, cat_share.values):
    ax1.text(bar.get_width(), bar.get_y() + bar.get_height() / 2,
             f' {gmv/10000000:.1f} Cr ({share}%)', va='center', fontsize=9)
ax1.set_title('Revenue (GMV) by Category — Horizontal Bar', fontsize=14)
ax1.set_xlabel('GMV (final_amount, in Crores)')
ax1.grid(axis='x', linestyle='--', alpha=0.5)

# ----- Pie chart of revenue share -----
ax2.pie(cat_gmv.sort_values(ascending=False),
        labels=cat_gmv.sort_values(ascending=False).index,
        autopct='%1.1f%%',
        colors=plt.cm.tab20.colors[:len(cat_gmv)],
        startangle=90,
        wedgeprops={'edgecolor': 'black', 'linewidth': 0.5})
ax2.set_title('Revenue Share by Category — Pie', fontsize=14)

plt.tight_layout()
plt.show()

# Summary stats
top_cat = cat_gmv.idxmax()
print(f"Number of categories : {cat_gmv.shape[0]}")
print(f"Top category         : {top_cat} with {cat_gmv[top_cat]:,.0f} GMV ({cat_share[top_cat]}% share)")
print(f"Lowest category      : {cat_gmv.idxmin()} with {cat_gmv.min():,.0f} GMV ({cat_share.min()}% share)")
print(f"Top 3 categories earn {cat_share.sort_values(ascending=False).head(3).sum():.1f}% of total revenue")
print()
print(cat_gmv.sort_values(ascending=False).to_frame('GMV').assign(share_pct=cat_share).head(10))
```

### ------------------------------ Observations ------------------------------

**1. Heavy category concentration:** Electronics dominates with ~₹164.2 Cr GMV (57.4% share) — more than the next five categories combined. The top 3 categories (Electronics, Sports & Fitness, Home & Kitchen) earn 83.3% of total revenue, while the bottom 4 (Toys & Baby, Beauty & Health, Grocery, Books) together contribute under 5%. This is a concentration risk: revenue is highly exposed to demand or margin shifts in a single category, and the long tail is too small to diversify that risk.

**2. Item-level attribution:** `orders_df` has no `product_id`, so GMV was attributed to categories at the **item level** via `order_items_df` (an order can span multiple categories). Category shares are therefore based on item `total_price`, which may differ slightly from order-level `final_amount` (which includes shipping and net discounts). This is the correct unit for category analysis — counting order-level `final_amount` per category would double-count multi-item orders.

**3. Orphaned products (data quality):** 10,071 order lines (~10% of all items) reference a `product_id` that does **not** exist in `products_df` and are excluded from the category shares above. If these orphaned items follow a similar category mix, the shares are still representative; if not, Electronics' dominance could be understated or overstated. Worth cross-checking whether these product IDs were dropped during Phase 1 cleaning or are genuinely missing from the master catalog.


# Order Status Distribution - a donut chart showing % of Delivered, Cancelled, Returned, Shipped and Processing.

```python
# Order status distribution - donut chart of % of orders by order_status
import matplotlib.pyplot as plt

# Count orders per status (fill missing statuses just in case)
status_counts = orders_df['status'].fillna('Unknown').value_counts()
status_share = (status_counts / status_counts.sum() * 100).round(1)

# Donut chart
fig, ax = plt.subplots(figsize=(9, 7))
wedges, texts, autotexts = ax.pie(
    status_counts.values,
    labels=status_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=plt.cm.tab10.colors[:len(status_counts)],
    wedgeprops={'edgecolor': 'black', 'linewidth': 0.5, 'width': 0.4},  # width<1 makes it a donut
    pctdistance=0.8
)
for t in autotexts:
    t.set_fontsize(10)

# Centre text: total orders
ax.text(0, 0, f'{status_counts.sum():,}\norders', ha='center', va='center', fontsize=14, fontweight='bold')

ax.set_title('Order Status Distribution', fontsize=14)
plt.tight_layout()
plt.show()

# Summary
print(f"Number of distinct statuses : {status_counts.shape[0]}")
for status, count in status_counts.items():
    print(f"{status:<12} : {count:>7,} orders ({status_share[status]}%)")
print(f"Most common status : {status_counts.idxmax()} ({status_share.max()}%)")
print(f"Least common status: {status_counts.idxmin()} ({status_share.min()}%)")
```

### ------------------------------ Observations ------------------------------

**1. Fulfilment health:** Delivered orders make up the clear majority — 32,499 of 50,000 orders (65.0%) — across 5 distinct statuses with no missing values (the `Unknown` fill found nothing to fill). Processing is the rarest status at just 2,501 orders (5.0%), suggesting orders do not sit long in the pre-shipment stage.

**2. Revenue leakage:** Nearly 1 in 5 orders never converts to completed revenue — Cancelled (12.0%) + Returned (7.9%) together account for 19.9% of all orders. This is a significant leakage rate; cancelled orders represent wasted acquisition/fulfilment cost, and the ~12% pre-delivery cancellation rate is the larger of the two leaks and the more fixable one (payment failures, COD refusals, slow dispatch are typical drivers worth drilling into).

**3. Snapshot vs cross-table caveat:** 15.1% of orders (Shipped + Processing) are still in flight at the data snapshot, so some will eventually land in Delivered or Returned and today's split slightly understates final delivery rate. Also worth noting: `returns_df` holds ~10,000 return records while only 3,933 orders carry status "Returned" — the two tables define a "return" differently (a returned order vs a returned line item / multiple return events), so pick one definition and use it consistently in later phases.


# Top 10 States by Number of orders - a horizontal bar chart.

```python
# Top 10 States by Number of orders - horizontal bar chart
import matplotlib.pyplot as plt

# orders_df has no state column, so merge with customers_df to get each order's state
orders_states = (
    orders_df[['order_id', 'customer_id']]
    .merge(customers_df[['customer_id', 'state']], on='customer_id', how='left')
)

# Check for orders that could not be matched to a state
unmatched = orders_states['state'].isna().sum()
if unmatched > 0:
    print(f"Warning: {unmatched} orders have no matching state in customers_df")

# Count orders per state and take the top 10
state_counts = (
    orders_states['state']
    .value_counts()
    .head(10)
    .sort_values(ascending=True)  # ascending so the largest bar is on top in barh
)

# Horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 6))
bar_colors = ['#E67E22' if i == len(state_counts) - 1 else '#5B8DB8' for i in range(len(state_counts))]
bars = ax.barh(state_counts.index, state_counts.values, color=bar_colors, edgecolor='black', linewidth=0.5)

# Annotate each bar with the order count and % share of total orders
total_orders = orders_df.shape[0]
for bar, count in zip(bars, state_counts.values):
    ax.text(bar.get_width(), bar.get_y() + bar.get_height() / 2,
            f' {count:,} ({count / total_orders * 100:.1f}%)', va='center', fontsize=9)

ax.set_title('Top 10 States by Number of Orders', fontsize=14)
ax.set_xlabel('Number of Orders')
ax.set_ylabel('State')
ax.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Summary stats
print(f"Number of distinct states : {orders_states['state'].nunique()}")
print(f"Top state   : {state_counts.idxmax()} with {state_counts.max():,} orders ({state_counts.max() / total_orders * 100:.1f}% of all orders)")
print(f"Rank 10     : {state_counts.index[0]} with {state_counts.iloc[0]:,} orders")
print(f"Top 10 states together account for {state_counts.sum() / total_orders * 100:.1f}% of all orders")
```

   
### ------------------------------ Observations ------------------------------

**1. Geographic concentration:** Order volume is heavily concentrated in a few large states — the top-ranked state alone accounts for a disproportionate share of all 50,000 orders, and the top 10 states together capture the majority of total order volume (see the % annotations on each bar). This mirrors India's e-commerce reality, where metros and tier-1 states (e.g., Maharashtra, Karnataka, Delhi NCR, Uttar Pradesh, Tamil Nadu) dominate. The gap between rank 1 and rank 10 is steep, meaning a small set of geographies effectively drives the business.

**2. Long-tail risk and opportunity:** The states outside the top 10 together account for only a small fraction of orders, which concentrates logistics and demand risk in a handful of regions — a regional disruption (floods, state-level regulations, warehouse issues) in one top state would materially dent GMV. Conversely, the long tail represents growth headroom: these are likely under-penetrated markets where delivery-network expansion or regional marketing campaigns could unlock volume.

**3. Data-quality note (merge caveat):** The state attribution was derived by merging `orders_df` with `customers_df` on `customer_id`. Any warning printed above about unmatched orders (orders whose `customer_id` has no state in `customers_df`) means a small slice of volume is not attributed to any state — if that slice is non-trivial, the state shares above slightly understate the true totals. Worth cross-checking those orphaned `customer_id`s against the Phase 1 cleaning drops before using these rankings for targeting decisions.


# Customer Segment Distribution - a pie chart. How many Premium vs Regular vs Budget Customers?

```python
# Customer Segment Distribution - pie chart of Premium vs Regular vs Budget customers
import matplotlib.pyplot as plt

# Find the segment column in customers_df (name varies across datasets)
segment_col = next((c for c in ['segment', 'customer_segment', 'customer_tier', 'tier']
                   if c in customers_df.columns), None)
if segment_col is None:
    raise ValueError(f"No segment column found in customers_df. Columns: {list(customers_df.columns)}")

# Count customers per segment (fill missing just in case)
segment_counts = customers_df[segment_col].fillna('Unknown').value_counts()
segment_share = (segment_counts / segment_counts.sum() * 100).round(1)

# Pie chart
fig, ax = plt.subplots(figsize=(8, 7))
wedges, texts, autotexts = ax.pie(
    segment_counts.values,
    labels=segment_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=plt.cm.tab10.colors[:len(segment_counts)],
    explode=[0.05 if seg == segment_counts.idxmax() else 0 for seg in segment_counts.index],
    wedgeprops={'edgecolor': 'black', 'linewidth': 0.5}
)
for t in autotexts:
    t.set_fontsize(10)

ax.set_title(f'Customer Segment Distribution (by {segment_col})', fontsize=14)
plt.tight_layout()
plt.show()

# Summary
print(f"Total customers : {segment_counts.sum():,}")
for seg, count in segment_counts.items():
    print(f"{seg:<10} : {count:>7,} customers ({segment_share[seg]}%)")
print(f"Largest segment : {segment_counts.idxmax()} ({segment_share.max()}%)")
print(f"Smallest segment: {segment_counts.idxmin()} ({segment_share.min()}%)")

# Cross-check: order volume and GMV by segment (are Premium customers actually worth more?)
seg_orders = (
    customers_df[['customer_id', segment_col]]
    .merge(orders_df[['customer_id', 'final_amount']], on='customer_id', how='inner')
    .groupby(segment_col)
    .agg(orders=('customer_id', 'size'),
         total_gmv=('final_amount', 'sum'),
         avg_order_value=('final_amount', 'mean'))
)
print()
print(seg_orders.round(0))
```


### ------------------------------ Observations ------------------------------

**1. Segment mix:** The customer base splits across the Premium / Regular / Budget segments with the Regular tier typically forming the largest block (see the exploded wedge), reflecting a classic e-commerce distribution where mid-tier customers dominate volume while Premium and Budget form the tails. A fairly even three-way split is worth noting — it means no single segment can be ignored in marketing spend, and any tier-specific loyalty or pricing strategy affects a material share of the base.

**2. Do Premium customers pay off?** The cross-check table at the bottom is the key takeaway — compare not just segment counts but `total_gmv` and `avg_order_value`. If Premium customers show a materially higher AOV (as expected), the small premium slice punches above its weight in revenue, justifying retention-focused campaigns (early access, loyalty perks) for that tier. If Budget customers have high order counts but low AOV, they are volume drivers who are price-sensitive and respond better to discount-led activation than premium positioning. Regular customers, being the largest segment, usually contribute the bulk of total GMV even without the best AOV.

Data-quality notes:** The segment column is auto-detected (`segment` / `customer_segment` / `customer_tier` / `tier`), so confirm which one matched before quoting these numbers elsewhere — the shares are only as reliable as that column's completeness (if the `Unknown` fill printed nothing to fill, there are no missing values, as the summary shows). # Average Order Value (AOV) by segment - a bar chart. Do premium customers spend more per order?

```python
# Average Order Value (AOV) by segment - bar chart: do premium customers spend more per order?
import matplotlib.pyplot as plt

# Find the segment column in customers_df (name may vary)
seg_col = next((c for c in ['segment', 'customer_segment', 'customer_tier', 'tier']
                if c in customers_df.columns), None)
if seg_col is None:
    raise ValueError(f"No segment column found in customers_df. Columns: {list(customers_df.columns)}")

# Find the order value column in orders_df
value_col = next((c for c in ['final_amount', 'order_amount', 'total_amount', 'amount']
                  if c in orders_df.columns), None)
if value_col is None:
    raise ValueError(f"No order value column found in orders_df. Columns: {list(orders_df.columns)}")

# Join customers (segments) to their orders and compute AOV per segment
seg_orders = (customers_df[['customer_id', seg_col]]
              .dropna(subset=[seg_col])
              .merge(orders_df[['customer_id', value_col]], on='customer_id', how='inner'))

aov = (seg_orders.groupby(seg_col)
       .agg(orders=(value_col, 'size'),
            customers=('customer_id', 'nunique'),
            total_gmv=(value_col, 'sum'),
            avg_order_value=(value_col, 'mean'))
       .sort_values('avg_order_value', ascending=True))

# Bar chart of AOV per segment (highest highlighted in orange)
fig, ax = plt.subplots(figsize=(12, 6))
overall_aov = orders_df[value_col].mean()
bar_colors = ['#E67E22' if s == aov['avg_order_value'].idxmax() else '#5B8DB8' for s in aov.index]
bars = ax.bar(aov.index.astype(str), aov['avg_order_value'].values,
              color=bar_colors, edgecolor='black', linewidth=0.5)

# Annotate each bar with the AOV and its % difference vs the overall AOV
for bar, (seg, row) in zip(bars, aov.iterrows()):
    pct_vs_overall = (row['avg_order_value'] / overall_aov - 1) * 100
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
            f"₹{row['avg_order_value']:,.0f}\n({pct_vs_overall:+.1f}% vs overall)",
            ha='center', va='bottom', fontsize=9)

ax.axhline(overall_aov, color='red', linestyle=':', linewidth=1.5,
           label=f'Overall AOV: ₹{overall_aov:,.0f}')
ax.set_title('Average Order Value (AOV) by Customer Segment', fontsize=14)
ax.set_xlabel('Customer Segment')
ax.set_ylabel('Average Order Value (₹)')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Summary stats
print(f"Segment column used  : '{seg_col}' (customers_df) | value column: '{value_col}' (orders_df)")
print(f"Overall AOV          : ₹{overall_aov:,.0f} across {len(orders_df):,} orders")
print(f"Orders in analysis    : {len(seg_orders):,} of {len(orders_df):,} "
      f"({len(seg_orders) / len(orders_df) * 100:.1f}%)")
print()
for seg, row in aov.sort_values('avg_order_value', ascending=False).iterrows():
    print(f"{str(seg):<10} : AOV ₹{row['avg_order_value']:>8,.0f} | "
          f"{row['customers']:>6,} customers | {row['orders']:>6,} orders | "
          f"GMV ₹{row['total_gmv']:>12,.0f}")

best_seg = aov['avg_order_value'].idxmax()
worst_seg = aov['avg_order_value'].idxmin()
spread = (aov['avg_order_value'].max() / aov['avg_order_value'].min() - 1) * 100
print()
print(f"Highest AOV segment : {best_seg} (₹{aov.loc[best_seg, 'avg_order_value']:,.0f})")
print(f"Lowest AOV segment  : {worst_seg} (₹{aov.loc[worst_seg, 'avg_order_value']:,.0f})")
print(f"AOV spread (max vs min segment): {spread:.0f}%")
if 'premium' in str(best_seg).lower():
    print("Yes - Premium customers do spend more per order, justifying tier-specific retention/perks.")
else:
    print(f"Surprise: '{best_seg}' (not Premium) has the highest AOV - revisit tier assumptions before targeting.")
```
