# For each table run a basic profile check: row count, column count and data types.

```python
def basic_profile_check(df, name="Dataframe"):
    # creating a list that stores the column names of the dataframe
    df_cols = df.columns.tolist()

    print(f"{'-'* 100}")
    
    # printing the number of rows for the dataframe
    print(f"{name} has {len(df)} rows")

    print(f"{'-'* 100}")
    
    # printing the number of columns for the dataframe
    print(f"{name} has {len(df_cols)} columns")

    print(f"{'-'* 100}")
    
    # printing the column datatypes
    for col in df_cols:
        print(f"{col} is of type {df[col].dtype}")
    
    
```

```python
print("Basic Profile Check for Inventory Dataframe")
basic_profile_check(inventory_df, "inventory.csv")
```

```python
print("Basic Profile Check for Customers Dataframe")
basic_profile_check(customers_df, "customers.csv")
```

```python
print("Basic profile check for Order Items Dataframe")
basic_profile_check(order_items_df, "order_items.csv")
```

# Check for missing values in every column - report which columns have nulls and what % of columns are affected.

```python
def checking_missing_values(df, name="data_frame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)
    null_rows = False

    for col in cols:
        null_values = df[col].isnull().sum()
        if null_values > 0:
            null_rows = True
            percentage_rows_affected = round((null_values / len(df)) * 100.00, 2)
            print(f'{col} in {name} has {null_values} null values which affects {percentage_rows_affected}% of rows.')
    
    if null_rows is False:
        print(f'{name} has 0 null values.')
```

```python
checking_missing_values(customers_df, "customers.csv")
```

```python
checking_missing_values(inventory_df, "inventory.csv")
```

```python
checking_missing_values(order_items_df, "order_items.csv")
```

```python
checking_missing_values(orders_df, "orders.csv")
```

```python
checking_missing_values(payments_df, "payments.csv")
```

```python
checking_missing_values(products_df, "products.csv")
```

```python
checking_missing_values(returns_df, "returns.csv")
```

```python
checking_missing_values(suppliers_df, "suppliers.csv")
```
